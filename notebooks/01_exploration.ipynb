## Fichier postings.csv

```python
import pandas as pd
import matplotlib.pyplot as plt

# Charger les données
df = pd.read_csv("../data/postings.csv")

# Informations générales
print("Dimensions :", df.shape)
print("\nColonnes :")
print(df.columns)

print("\nInformations :")
df.info()

print("\nAperçu :")
display(df.head())

print("\nStatistiques numériques :")
display(df.describe())

print("\nValeurs uniques :")
display(df.nunique())

# Visualisation : Top 10 entreprises
df["company_name"].value_counts().head(10).plot(
    kind="bar",
    figsize=(10,5),
    title="Top 10 entreprises"
)
plt.show()
```
